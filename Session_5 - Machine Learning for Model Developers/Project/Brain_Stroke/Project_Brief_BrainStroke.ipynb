# Capstone Project: Brain Stroke Prediction (Classification)

*A competition-style capstone covering the full data science workflow on a classification problem.*

## Your task

Implement the **full data science process (up to model evaluation)** on the brain stroke dataset. You will explore and clean the data, analyse it, build and compare classification models, evaluate them, and predict on a held-out **unseen** set that the training team scores on a **leaderboard**. Keep careful notes of every decision.

You are required to:

- Perform **extensive EDA** (data quality, distributions, relationships), using your pandas, data-quality and statistics skills.
- Implement, **compare and contrast** the classification models learned in the module.
- Document the full process (up to model evaluation) with clear markdown and code comments.
- Justify **how the models were tested** and **how the parameters were chosen**.
- Produce an overall **performance report**: classification report, confusion matrix, accuracy, precision, recall and F1.
- Finish with **conclusions, lessons learned and future work**.

Keep notes of what you drop, what you encode, what you train on, what you leave behind, and your assumptions. **Your model is evaluated by the training team on the unseen data.**

## The dataset

- **Dataset:** `brain_stroke_data.csv` (with the target) | **Unseen:** `brain_stroke_unseen.csv` (no target)
- **Target:** `stroke` (1 = had a stroke, 0 = did not). **Imbalanced** (about 5% positive).
- **Source:** [Kaggle: Brain Stroke Dataset](https://www.kaggle.com/datasets/jillanisofttech/brain-stroke-dataset).

| Column | Description |
|---|---|
| `gender` | Male / Female |
| `age` | Age in years |
| `hypertension` | 0 = no, 1 = yes |
| `heart_disease` | 0 = no, 1 = yes |
| `ever_married` | Yes / No |
| `work_type` | Type of employment (Private, Self-employed, Govt_job, children, etc.) |
| `Residence_type` | Urban / Rural |
| `avg_glucose_level` | Average blood glucose level |
| `bmi` | Body mass index |
| `smoking_status` | formerly smoked / never smoked / smokes / Unknown |
| `stroke` | **Target.** 1 = had a stroke, 0 = did not |

## Competition and scoring

This project runs as a **competition**. You get the labelled data to train on and an **unseen set with no target**. You predict `stroke` for every row of the unseen set and submit a CSV. The training team scores every submission against a **hidden answer key you never see** and publishes a leaderboard.

- **Leaderboard metric:** F1 on the stroke class. The data is imbalanced (~5% positive), so accuracy is not used (predicting "no stroke" for everyone already scores about 95%). The goal is to catch real strokes without too many false alarms.
- **Submission format:** a CSV with a single column `Predictions`, one row per unseen record in the original order. Name it `name_surname_brain_stroke.csv`.

## Deliverables

- **Jupyter notebook:** the full process, documented (markdown + code comments), cleaned of throwaway exploratory code.
- **Predictions CSV:** one column named `Predictions`, one row per record in `brain_stroke_unseen.csv` (same order), named `name_surname_brain_stroke.csv`.

**Submission:** details will be provided by the training team.
