# Capstone Project: Airplane Crash Fatalities (Regression)

*A competition-style capstone covering the full data science workflow on a regression problem that needs heavy preprocessing.*

## Your task

Implement the **full data science process (up to model evaluation)** on the airplane crashes dataset. You will explore and clean the data, engineer features, analyse it, build and compare regression models, evaluate them, and predict on a held-out **unseen** set that the training team scores on a **leaderboard**. Keep careful notes of every decision.

You are required to:

- Perform **extensive EDA** (data quality, distributions, relationships), using your pandas, data-quality and statistics skills.
- Implement, **compare and contrast** the regression models learned in the module.
- Document the full process (up to model evaluation) with clear markdown and code comments.
- Justify **how the models were tested** and **how the parameters were chosen**.
- Produce an overall **performance report**: R-squared, RMSE, MSE and MAE.
- Finish with **conclusions, lessons learned and future work**.

Keep notes of what you drop, what you encode, what you train on, what you leave behind, and your assumptions. **Your model is evaluated by the training team on the unseen data.**

## The dataset

- **Dataset:** `airplane_crashes_data.csv` (with the target) | **Unseen:** `airplane_crashes_unseen.csv` (no target)
- **Target:** `Fatalities` (number of people killed on board).
- **Source:** [Kaggle: Airplane Crashes and Fatalities](https://www.kaggle.com/datasets/thedevastator/airplane-crashes-and-fatalities).
- **Note:** this dataset **requires considerable preprocessing** (dates as text, free-text fields, very high-cardinality columns, missing values), so feature engineering is most of the work.

| Column | Description |
|---|---|
| `Date`, `Time` | Date (MM/DD/YYYY) and time of the crash (both need parsing; `Time` is often missing) |
| `Location`, `Route` | Free-text location and route |
| `Operator` | Airline or operator (very high cardinality) |
| `Flight #`, `Registration`, `cn/In` | Identifiers (mostly missing / high cardinality) |
| `Type` | Aircraft type (very high cardinality) |
| `Aboard` | Number of people on board (a strong predictor) |
| `Ground` | People killed on the ground |
| `Summary` | Free-text narrative of the crash |
| `Fatalities` | **Target.** Number of people killed on board |

## Competition and scoring

This project runs as a **competition**. You get the labelled data to train on and an **unseen set with no target**. You predict `Fatalities` for every row of the unseen set and submit a CSV. The training team scores every submission against a **hidden answer key you never see** and publishes a leaderboard.

- **Leaderboard metric:** RMSE (lower is better); R-squared and MAE are also reported.
- **Submission format:** a CSV with a single column `Predictions`, one row per unseen record in the original order. Name it `name_surname_airplane_crashes.csv`. Remember fatalities cannot be negative.

## Challenge (optional extension): mining the free-text `Summary`

The dataset includes a free-text `Summary` (a short narrative of each crash) that most solutions simply drop. As an optional extension, turn it into numeric features and test whether they improve your regression. Words like "exploded", "collision", "engine failure" or "sea" may carry signal about how deadly a crash was.

Text-handling techniques to try:

- **Bag of words** (`CountVectorizer`) or **TF-IDF** (`TfidfVectorizer`) to vectorise the narrative into numeric columns.
- **Stemming / lemmatisation** (for example with NLTK) to collapse word variants ("crashed / crashing / crash") into one token before vectorising.
- **Stop-word removal, n-grams and `max_features`** to keep the vocabulary manageable and focused.

Combine the text features with your numeric features (a `ColumnTransformer` is handy), compare the held-out RMSE with and without them, and remember to apply the same fitted vectoriser to the unseen set before predicting.
