## Data source and data dictionary

**Source:** UCI Machine Learning Repository, [Default of Credit Card Clients](https://archive.ics.uci.edu/dataset/350/default+of+credit+card+clients). Original study: Yeh, I. C., and Lien, C. H. (2009), "The comparisons of data mining techniques for the predictive accuracy of probability of default of credit card clients," *Expert Systems with Applications*, 36(2), 2473-2481. It covers credit-card clients in Taiwan from April to September 2005. Our copy dropped the `ID` column, renamed the target to `default`, and split the data into training and unseen sets.

| Column | Description |
|---|---|
| `LIMIT_BAL` | Amount of credit granted (NT dollars), including own and family/supplementary credit |
| `SEX` | Gender (1 = male, 2 = female) |
| `EDUCATION` | Education (1 = graduate school, 2 = university, 3 = high school, 4 = other; 0, 5, 6 undocumented, treated as "other") |
| `MARRIAGE` | Marital status (1 = married, 2 = single, 3 = other; 0 undocumented, treated as "other") |
| `AGE` | Age in years |
| `PAY_0`, `PAY_2` ... `PAY_6` | Repayment status in September, August, ..., April 2005 (-1 = paid duly, 1 to 9 = months late; -2 and 0 undocumented, usually no consumption or revolving). `PAY_0` is the most recent month. |
| `BILL_AMT1` ... `BILL_AMT6` | Bill statement amount (NT dollars) for September ... April 2005 |
| `PAY_AMT1` ... `PAY_AMT6` | Amount of previous payment (NT dollars) for September ... April 2005 |
| `default` | **Target.** Did the client default on next month's payment? (1 = yes, 0 = no) |
