## Scenario (baseline): Credit card default (Classification)

- **Dataset:** `credit_default_data.csv` (27,000 clients, target included)
- **Unseen data:** `credit_default_unseen.csv` (3,000 clients, no target)
- **Target variable:** `default` (1 = the client defaulted next month, 0 = did not)
- **Data source:** UCI "Default of Credit Card Clients" (Taiwan). Features cover the credit limit (`LIMIT_BAL`), demographics (`SEX`, `EDUCATION`, `MARRIAGE`, `AGE`), the six most recent repayment statuses (`PAY_0`, `PAY_2`...`PAY_6`), bill amounts (`BILL_AMT1`...`6`) and payment amounts (`PAY_AMT1`...`6`).
- **Notes:** the classes are **imbalanced** (about 22% default). Several categorical fields are stored as integer codes and contain **undocumented values** (for example `EDUCATION` has 0, 5 and 6; `MARRIAGE` has 0; `PAY_*` has -2 and 0). Deciding how to handle these is part of the task.

## Data source and data dictionary

**Source:** UCI Machine Learning Repository, ["Default of Credit Card Clients"](https://archive.ics.uci.edu/dataset/350/default+of+credit+card+clients). Original study: Yeh, I. C., and Lien, C. H. (2009), "The comparisons of data mining techniques for the predictive accuracy of probability of default of credit card clients," *Expert Systems with Applications*, 36(2), 2473-2481. The data covers credit-card clients in Taiwan from April to September 2005.

*Our copy has been lightly adapted: the `ID` column was dropped, the target renamed to `default`, and the data split into a training set (`credit_default_data.csv`, with the target) and an unseen validation set (`credit_default_unseen.csv`, with the target withheld for the competition).*

| Column | Description |
|---|---|
| `LIMIT_BAL` | Amount of credit granted (NT dollars), including the individual's own and any family/supplementary credit |
| `SEX` | Gender (1 = male, 2 = female) |
| `EDUCATION` | Education level (1 = graduate school, 2 = university, 3 = high school, 4 = other; codes 0, 5, 6 are undocumented and treated as "other") |
| `MARRIAGE` | Marital status (1 = married, 2 = single, 3 = other; code 0 is undocumented and treated as "other") |
| `AGE` | Age in years |
| `PAY_0`, `PAY_2` ... `PAY_6` | Repayment status in September, August, ..., April 2005 respectively (-1 = paid duly, 1 = one month late, 2 = two months late, up to 9 = nine or more months late; codes -2 and 0 are undocumented, usually taken as no consumption or revolving credit). `PAY_0` is the most recent month. |
| `BILL_AMT1` ... `BILL_AMT6` | Bill statement amount (NT dollars) for September ... April 2005 |
| `PAY_AMT1` ... `PAY_AMT6` | Amount of the previous payment (NT dollars) for September ... April 2005 |
| `default` | **Target.** Did the client default on next month's payment? (1 = yes, 0 = no) |

## Competition and scoring

This project runs as a **competition**. You are given `credit_default_data.csv` (with the `default` target, to train on) and `credit_default_unseen.csv` (the **validation set, with no target**). You predict `default` for every row of the validation set and submit your predictions. The training team scores every submission against a **hidden answer key that you never see**, and publishes a **leaderboard**.

- **Leaderboard metric:** F1 on the defaulting class. The data is imbalanced, so accuracy is not used (predicting "no default" for everyone already scores about 78%). The goal is to catch real defaulters without too many false alarms.
- **Submission format:** a CSV with a single column `Predictions`, one row per validation record in the original order, named `name_surname_credit_default.csv`.
- Optionally, submit predicted probabilities (values between 0 and 1) instead of 0/1 labels, and the team will also report ROC AUC.
