# Section 4: List comprehensions

List comprehensions are a concise, Pythonic way to **build a new list from an existing iterable**. They replace many simple `for` loops with a single readable line (Session 1, notebook 01.02, Section 9).

**Basic syntax:**

```python
new_list = [expression for item in iterable]
```

**With a condition:**

```python
new_list = [expression for item in iterable if condition]
```

### From loop to comprehension

The loop from Section 3 becomes one line. All three versions (manual, loop, comprehension) produce the same result; the comprehension is **more concise**, **faster** (Python optimises it internally), and **more readable** once you know the pattern.

````python
# Best: list comprehension -- one line
````

# Section 7: Objects, methods and attributes

Everything in Python is an **object**. Object-oriented programming models the world as objects, each combining (Session 1, notebook 01.04):

- **Attributes**: data the object holds.
- **Methods**: actions the object can perform.

A **class** is the blueprint; an **instance** (object) is a specific realisation of it. You rarely need to write your own classes in data work, but you **use** objects all the time: a string, a list and a pandas DataFrame are all objects.

To access an attribute or a method, put a dot after the object (`obj.attribute`). When the attribute is callable it is a **method** and you call it with brackets; otherwise it is just a piece of data about the object (Session 1, notebook 01.01). A method is a function that belongs to an object: `len(word)` is a function, `word.upper()` is a method. The built-in `dir()` lists an object's attributes and methods.

| Method | Description | Example |
|--------|-------------|----------|
| `.upper()` | All uppercase | `'hello'.upper()` gives `'HELLO'` |
| `.lower()` | All lowercase | `'HELLO'.lower()` gives `'hello'` |
| `.title()` | Each word capitalised | `'hello world'.title()` gives `'Hello World'` |
| `.strip()` | Remove leading/trailing whitespace | `'  hi  '.strip()` gives `'hi'` |
| `.replace(old, new)` | Replace substring | `'cat'.replace('c','b')` gives `'bat'` |
| `.split(sep)` | Split into list | `'a,b,c'.split(',')` gives `['a','b','c']` |

# Section 9: Handling errors (try / except)

When Python meets a problem it cannot continue from, it **raises an exception**. If the exception is not handled, the program stops and prints a traceback. `try` / `except` lets you handle it instead (Session 1, notebook 01.05):

1. Python runs the `try` block.
2. If anything goes wrong, it jumps immediately to `except`.
3. If nothing goes wrong, `except` is skipped.
4. Execution continues after the `try` / `except`.

The exceptions you will see most often:

| Exception | When it occurs | Example |
|-----------|---------------|--------|
| `TypeError` | Wrong type for an operation | `'a' + 1` |
| `ValueError` | Right type, invalid value | `int('hello')` |
| `ZeroDivisionError` | Dividing by zero | `10 / 0` |
| `IndexError` | Sequence index out of range | `lst[99]` on a 3-element list |
| `KeyError` | Dictionary key not found | `d['missing']` |
| `FileNotFoundError` | File does not exist | `pd.read_csv('missing.csv')` |

Catching **specific** exception types lets you respond differently to different problems and, crucially, **avoids hiding bugs you did not expect**.

**Best practices:**

| Practice | Why |
|----------|-----|
| **Catch specific exceptions** | A bare `except:` hides real bugs |
| **Keep `try` blocks short** | Only wrap the code that might raise |
| **Write helpful messages** | Include the bad value and what was expected |
| **Don't silence exceptions** | `except: pass` is almost always wrong |
