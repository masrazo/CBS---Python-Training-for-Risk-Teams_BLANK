## About the data

Bikeshare schemes let people hire a bike from one docking station and return it at
another. The schemes in this dataset are among the best known public bikeshare open-data
programmes in the world, and the data is widely used for teaching and analytics:

- **Bay Wheels** : San Francisco Bay Area (operated by Lyft; formerly Ford GoBike).
- **Bluebikes** : Metropolitan Boston.
- **Capital Bikeshare** : Washington, D.C. metro area.
- **Divvy** : Chicago.
- **Santander Cycles** : London, UK (popularly known as "Boris Bikes").

Each scheme publishes anonymised **trip records** as open data. Broadly, **one row is one
trip**, and typically includes the start and end timestamps, the start and end stations,
a trip duration, and a rider category (for example member versus casual, or subscriber
versus customer). The exact column names and formats differ between systems and between
years, which is part of the challenge: do not assume a schema, inspect it.

Alongside the trips, each system has a **stations** file with station metadata (id, name,
latitude, longitude, and often capacity), and there is a shared `distances.csv` with
precomputed station-to-station distances you can use for enrichment.

The original open-data sources are listed in **Appendix B**, and the full list of files
available to you, with their S3 paths, is in **Appendix A**.
