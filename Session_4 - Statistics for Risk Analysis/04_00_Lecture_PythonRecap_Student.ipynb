### Anatomy of a function

- `def` starts the definition; give the function a **descriptive name**.
- **Parameters** go in the parentheses.
- The **docstring** (triple-quoted text) says what it does. Good documentation is not optional in shared code.
- The **body** does the work; `return` hands back the result.

Docstring and style references: [PEP 257 (docstrings)](https://peps.python.org/pep-0257/), [PEP 8 (style)](https://peps.python.org/pep-0008/), [Google Python style guide](https://google.github.io/styleguide/pyguide.html#38-comments-and-docstrings).

## Key Takeaways

| Tool | What it is for |
|---|---|
| list / dict / tuple / set | the core containers; list and dict are your daily drivers |
| `if` / `elif` / `else`, ternary | branch on the data; band values and handle special cases |
| `for` / `while` loop | run code once per item, or until a condition ends |
| list comprehension | build a list in one readable line |
| `enumerate`, `zip`, f-string, `lambda` | everyday convenience idioms |
| function + docstring | a named, documented, reusable step |
| composition | small functions snapped into a pipeline |
| NumPy array | vectorised maths and simple statistics, no loops |
| pandas DataFrame | load, inspect, select, filter, group, visualise, engineer columns |
