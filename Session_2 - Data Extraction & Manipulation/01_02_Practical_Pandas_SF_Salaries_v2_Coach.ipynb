### Exercise 5.6: Several stats per year at once

### Exercise 5.7: Top-paying job titles

Among job titles held by at least 50 people, which 5 have the highest average `TotalPayBenefits`?

> *Hint: First find the titles with >= 50 holders, filter to them with `.isin`, then group and sort.*

```python
# Your turn. Write your solution here:
```

```python
# Coach answer
counts = sal['JobTitle'].value_counts()
common = counts[counts >= 50].index
(
    sal[sal['JobTitle'].isin(common)]
    .groupby('JobTitle')['TotalPayBenefits'].mean()
    .sort_values(ascending=False)
    .head(5)
)
```

# Part 6: A small join (comparing pay to a yearly benchmark)

This dataset is a single table, so there's no second table to join to out of the box. But joins are everywhere in real work, so here's a realistic one. We'll bring in a tiny **reference table** holding one external figure per year, the approximate **San Francisco median household income**, and join it on `Year`. That lets us ask how city salaries compared to a typical household's income each year. Run the lookup cell first.

```python
# A small reference table: one external figure per year.
# (Approximate SF median household income, in dollars.)
benchmark = pd.DataFrame({
    'Year': [2011, 2012, 2013, 2014],
    'sf_median_income': [71304, 73012, 75604, 78378],
})
benchmark
```

### Exercise 6.1: Join the benchmark onto the salaries

Merge `benchmark` onto `sal` with a left join on `Year`. Call the result `sal_bench` and confirm it now has an `sf_median_income` column.

> *Hint: `pd.merge(left, right, on='Year', how='left')`.*

```python
# Your turn. Write your solution here:
```

```python
# Coach answer
sal_bench = pd.merge(sal, benchmark, on='Year', how='left')
sal_bench[['EmployeeName', 'Year', 'TotalPayBenefits', 'sf_median_income']].head()
```

### Exercise 6.2: Compare each salary to the benchmark

Add a column `pay_vs_median` = `TotalPayBenefits / sf_median_income`, then show the average ratio per `Year`. A value of 2.0 would mean the average city salary was twice a typical household's income that year.

> *Hint: divide the two columns to build the ratio, then `groupby('Year')` and take the mean.*

```python
# Your turn. Write your solution here:
```

```python
# Coach answer
sal_bench['pay_vs_median'] = sal_bench['TotalPayBenefits'] / sal_bench['sf_median_income']
sal_bench.groupby('Year')['pay_vs_median'].mean().round(2)
```

### Exercise 7.2: Export a yearly summary
