# Section 1: Container Types — Overview
---


So far, we’ve only worked with numbers and strings, learning how to write simple expressions with these data types. However, writing programs often involves managing more complex collections of data. This means you need to think about data structures for **storing data and algorithms** for manipulating it. In this section, we’ll explore some of Python’s powerful built-in data structures: **lists**, **tuples**, **sets**, and **dictionaries**.

Data Structures Overview:
- **String** **`str`**
  * **Immutable**: Once created, their contents cannot be changed
  * **Indexed by integers**: You access individual characters using integer indices
  * **Order**: The characters are stored in the order they were added

- **List** **`list`**
  * **Mutable**: You can modify, add, or remove items after the list is created
  * **Indexed by integers**: You access items using integer indices
  * **Order**: Items are stored in the order they were added
  * `[3, 5, 6, 3, 'dog', 'cat', False]`

- **Tuple** **`tuple`**
  * **Immutable**: Once created, their contents cannot be changed
  * **Indexed by integers**: You access items using integer indices
  * **Order**: Items are stored in the order they were added
  * `(3, 5, 6, 3, 'dog', 'cat', False)`

- **Set** **`set`**
  * **Mutable**: You can add or remove elements after creation
  * **Unordered**: Elements are stored in an arbitrary order (not indexed)
  * **Unique Elements**: A set cannot contain duplicate values
  * **Efficient Membership Checking**: Checking if an element exists (`x in set`) is fast
  * `{'banana', 'cherry', 'apple', 'orange'}`

- **Dictionary** **`dict`**
  * **Mutable**: You can modify, add, or remove key-value pairs after the dictionary is created.
  * **Key-value pairs**: Items are stored as key-value pairs, where each key is unique.
  * **Indexed by immutable keys**: Keys can be strings, numbers, or other immutable types, but not lists or other mutable types.
  * **Order**: Items are NOT stored in the order they were added (although recent Python versions do preserve insertion order)
  * `{'name': 'Jane', 'age': 23, 'fav_foods': ['pizza', 'fruit', 'fish']}`


**Key Points to Remember**:
* When defining lists and tuples, use commas (,) to separate the individual items.
* When defining sets, use commas ( , ) to separate items, but remember that duplicate values are automatically removed.
* When defining dictionaries, use a colon (:) to separate keys from values and commas (,) to separate key-value pairs.


**Operations with Sequences**:
Strings, lists, and tuples are all sequence types, which means they share some common features. You can use the following operators:

* **`+:`** Concatenates two sequences
* **`*:`** Repeats a sequence
* **`+=:`** Adds elements to a sequence (only works for mutable types like lists)
* **`*=:`** Repeats the sequence in place

## 2.2 Indexing

In Python, indexing refers to accessing individual elements within a sequence (like a list, string, or tuple). Python follows a zero-based index system, meaning that the first element in a sequence is at index 0, the second element is at index 1, and so on.

This is similar to how we indexed characters in strings, but now we can apply it to lists, which hold multiple items.

Example of Indexing in a List:

In Python, you can also access elements in a sequence using reverse indexing. Instead of starting from 0 for the first element, reverse indexing starts from -1 for the last element.

This is particularly useful when you want to access elements from the end of a list, string, or tuple without knowing the total length.

## 2.3 Nested Lists


In this example, we have declared two lists, `x` and `y`, each containing its own set of data. These two lists can then be placed into another list, say `z`, where `z` will hold both `x` and `y` as its elements.

This is called a **nested list**, where one list is contained inside another. Nested lists are conceptually similar to arrays in other programming languages, and we'll explore arrays further later.

### Indexing
**Indexing** in nested lists can be tricky if you don’t fully understand how Python indexing works. Let’s break it down step by step to make it easier to grasp.

Consider a nested list `z` that looks like this:

`z = [['parsnip', 'broccoli'], ['carrot', 'potato']]`


To access the item 'broccoli' in this list:

First, at index 0 of `z`, we have the list `['parsnip', 'broccoli']`.

At index 1, we have another list `['carrot', 'potato']`.


So, to access 'broccoli', we first need to use `z[0]` to get the first list `['parsnip', 'broccoli']`. Then, within that list, we access the second element using `z[0][1]`.

## 2.4 Slicing



While **indexing** is used to access a single element within a list, **slicing** allows you to access a sequence of elements from a list, effectively "slicing" the list into a smaller portion.

**Slicing** lets you specify the start and end indices of the elements you want to extract from the list. It's written as `list[a:b]`, where:

* a is the index of the first element to include (inclusive)
* b is the index of the last element to include (exclusive)

## 2.7 Copying Lists

Be careful when copying lists — a simple assignment **does not create an independent copy**:

```python
original = [1, 2, 3]
copy = original        # NOT a copy — both point to the same list
copy.append(4)
print(original)        # [1, 2, 3, 4]  — original was modified!
```

Use `.copy()` or `list()` to make a true shallow copy:

```python
true_copy = original.copy()   # or: true_copy = list(original)
true_copy.append(99)
print(original)               # [1, 2, 3, 4]  — unchanged
print(true_copy)              # [1, 2, 3, 4, 99]
```

**Tuples** are a type of data structure in Python that are similar to lists, but with one key difference: once a tuple is created, its **elements cannot be modified, changed, or erased**. This makes tuples **immutable**, which is why they are often used for storing fixed collections of data that should not be altered.

To define a tuple, a variable is assigned to parenthesis () or tuple( ).

````python
tup = (['hi', 'how', 'are you'], 1, 2, True)
tup
````

````python
tuple('hello world'), tuple('hello world'.split())
````

### Tuple Unpacking

Tuples can be assigned to elements assuming the have the right length. You can assign tuple elements to individual variables in one line:

```python

# Useful for swapping variables
a, b = 1, 2
```

# Section 4: Sets
<hr style="border-top:1px dashed">

Sets are mainly used to eliminate repeated numbers in a sequence or list. It is also used to perform some standard set operations.

Sets are declared using the `set()` function, which will initialize an empty set, or using curly brackets `{}`. Also `set([sequence])` can be executed to declare a set with elements. Note that unlike lists, the elements of a set are not in a sequence and **cannot be accessed by an index**.

Notice for `set3` that elements 2 and 3 are repeated initially but only recorded once. This is because each element in a set is **distinct**.

Notice too that the **order** of a set is **not preserved**.

Note that although curly brackets can be used to create a set, **{}** itself is **NOT** a set but a dictionary (more on that next chapter).

In Python, control flow statements determine the order in which code executes. Unlike many other languages that use brackets `{}` or keywords like `begin` and `end`, Python relies on **indentation** to define blocks of code.

This means that the **amount of whitespace (spaces or tabs) at the beginning of a line is crucial**. Proper indentation makes Python code more readable but can be a common source of errors for beginners.

Below, we will explore the following conditional statements:

- `if` – Executes a block of code if a condition is true.
- `else` – Runs a block of code when all previous conditions are false.
- `elif` – Checks additional conditions if the previous ones are false.
- **Nested conditions** – Conditional statements inside other conditionals to check multiple levels of conditions.



## 6.1 `if` Statement  

The `if` statement allows us to execute a block of code only if a specified condition evaluates to `True`.  

**Syntax:**
```python
if condition:
    # code runs only when condition is True
```

Python uses **indentation** (4 spaces or 1 tab) to define code blocks. This is not optional.

## 6.2 `if-else` Statements  


The `if-else` statement allows us to execute one block of code if a condition is `True`, and a different block if the condition is `False`.  

### Syntax:  
```python
if condition:
    # Code block executed if the condition is True
else:
    # Code block executed if the condition is False
```
If the condition inside if is met, the first block runs; otherwise, the else block executes.

## 6.3 `elif` (Else-If) Statements  

The `elif` statement allows multiple conditions to be checked in sequence. If the first condition is `False`, Python moves to the next `elif` condition. If none of the conditions are met, the `else` block (if present) executes.  

### Syntax:  
```python
if condition_1:  
    # Code block executed if condition_1 is True
elif condition_2:  
    # Code block executed if condition_1 is False and condition_2 is True
else:  
    # Code block executed if all previous conditions are False
```

You can use multiple `elif` statements to handle different cases before reaching the final else.

## 6.4 Nested Conditions  

A nested `if` statement occurs when an `if` statement is placed inside another `if`, `if-elif`, or `if-else` block. This allows for more complex decision-making based on multiple conditions.

### Syntax:  
```python
if condition_1:  
    if condition_2:  
        # Code executes if both condition_1 and condition_2 are True  
    else:  
        # Code executes if condition_1 is True but condition_2 is False  
else:  
    # Code executes if condition_1 is False  
```


```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry permitted")
    else:
        print("ID required")
else:
    print("Under 18: entry not permitted")
```

### `enumerate()` and `zip()`

**`enumerate()`** gives you both the index and value when iterating:

```python
fruits = ['apple', 'banana', 'cherry']
for i, fruit in enumerate(fruits):
    print(f'{i}: {fruit}')
# 0: apple
# 1: banana
# 2: cherry
```

**`zip()`** iterates over two or more sequences simultaneously:

```python
names  = ['Alice', 'Bob', 'Carol']
scores = [85, 72, 91]
for name, score in zip(names, scores):
    print(f'{name}: {score}')
```

````python
# enumerate
fruits = ['apple', 'banana', 'cherry']
for i, fruit in enumerate(fruits, start=1):
    print(f'{i}. {fruit}')

print()

# zip
names  = ['Alice', 'Bob', 'Carol']
scores = [85, 72, 91]
for name, score in zip(names, scores):
    status = 'Pass' if score >= 40 else 'Fail'
    print(f'{name}: {score} ({status})')
````

# Section 9: List Comprehensions
---

List comprehensions are a concise, Pythonic way to **build a new list from an existing iterable**. They replace many simple `for` loops with a single readable line.

**Basic syntax:**

```python
new_list = [expression for item in iterable]
```

**With a condition:**

```python
new_list = [expression for item in iterable if condition]
```

---
## 9.1 From Loop to Comprehension

Let's see the progression. Suppose we want the **squares** of `[1, 2, 3, 4, 5]`.

**Step 1 -- Manual repetition (the worst way):**

````python
# Manual repetition -- inflexible, doesn't scale
nums = [1, 2, 3, 4, 5]
````

**Step 2 -- Iteration with a `for` loop:**

````python
# Better: a loop
nums = [1, 2, 3, 4, 5]
````

**Step 3 -- The comprehension version:**

````python
# Best: list comprehension -- one line
nums = [1, 2, 3, 4, 5]
````

All three produce the same result. The comprehension is **more concise**, **faster** (Python optimises it internally), and **more readable** once you know the pattern.

## 9.4 Nested Comprehensions

You can nest a comprehension to flatten a 2D list, or to build a 2D structure:

````python
# Flatten a matrix
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [val for row in matrix for val in row]
print(flat)   # [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Build a 3x3 multiplication table
table = [[i*j for j in range(1, 4)] for i in range(1, 4)]
for row in table:
    print(row)
````

> **Tip:** if your nested comprehension is hard to read, switch back to a regular nested loop. Comprehensions should make code clearer, not the opposite.

---
## 9.5 Dict and Set Comprehensions

The same syntax works for **dictionaries** (with `key: value`) and **sets**:

```python
# dict comprehension
{key_expr: value_expr for item in iterable}

# set comprehension
{expr for item in iterable}
```

````python
# Dict: word -> length
words = ['python', 'pandas', 'numpy', 'matplotlib']
word_lengths = {w: len(w) for w in words}
print(word_lengths)

# Dict: invert a mapping
ranks = {'gold': 1, 'silver': 2, 'bronze': 3}
by_rank = {v: k for k, v in ranks.items()}
print(by_rank)

# Set: unique first letters
first_letters = {w[0] for w in words}
print(first_letters)
````
