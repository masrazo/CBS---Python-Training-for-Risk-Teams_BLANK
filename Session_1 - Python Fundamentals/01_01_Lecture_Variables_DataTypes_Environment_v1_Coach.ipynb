In Python, everything is an **object**. This means that every value, whether it's a number, a string, or even a more complex structure like a list or a function, is treated as an object. Some of the basic types include:

  - `10`
  - `-3`
- **`float`** (**float**; a number that has a decimal place)
  - `7.41`
  - `-0.006`
- **`str`** (**string**; a sequence of characters enclosed in single quotes, double quotes, or triple quotes)
  - `'this is a string using single quotes'`
  - `"this is a string using double quotes"`
  - `'''this is a triple quoted string using single quotes'''`
  - `"""this is a triple quoted string using double quotes"""`
- **`bool`** (**boolean**; a binary value that is either true or false)
  - `True`
  - `False`
- **`NoneType`** (a special type representing the absence of a value)
  - `None`


To find out what kind of object a variable is pointing to, you can use the built-in function **`type()`**.
This function tells you the type of the object the variable refers to, whether it's an integer, string, list, or something else.

### Checking Types with `type()` and `isinstance()`

Use **`type()`** to find out what type an object is, and **`isinstance()`** to check if an object is an instance of a particular type.

````python
x = 2
y = 4.0
w = 'hello'
flag = True

# type()
print(type(x))    # <class 'int'>
print(type(y))    # <class 'float'>
print(type(w))    # <class 'str'>
print(type(flag)) # <class 'bool'>
````

````python
# isinstance() — returns True or False
print(isinstance(x, int))       # True
print(isinstance(y, float))     # True
print(isinstance(w, str))       # True
print(isinstance(flag, bool))   # True
print(isinstance(x, str))       # False
````

## 4.1 Type Casting

In Python, there may be situations where you want to explicitly specify the type of a variable. This can be achieved through **casting**, which is the process of converting one data type into another.

**Casting** in Python is done using constructor functions, which are **built-in functions** that create a specific type of object from another type. These functions allow you to convert between different types in a controlled way.

Here are the primary casting functions:

- `int()` - This function converts a value into an integer. It can be used with:

  * An integer (no change).
  * A float (it removes the decimal part).
  * A string (as long as the string represents an integer or a float).

- `float()` - This function converts a value into a float (a number with decimals). It can be used with:
  * An integer (converts it to a float).
  * A float literal (no change).
  * A string (as long as the string represents a number, either integer or float).
- `str()` - This function converts a value into a string. It can be used with:
  * Strings (no change).
  * Integers (converts the integer to its string representation).
  * Floats (converts the float to its string representation).

### Operator Precedence

Python follows standard mathematical precedence rules (similar to BIDMAS/BODMAS):

| Priority | Operator | Description |
|----------|----------|-------------|
| 1 (highest) | `**` | Exponentiation |
| 2 | `+x`, `-x` | Unary plus/minus |
| 3 | `*`, `/`, `//`, `%` | Multiplication, Division, Floor div, Modulo |
| 4 | `+`, `-` | Addition, Subtraction |
| 5 | `==`, `!=`, `<`, `>`, `<=`, `>=` | Comparisons |
| 6 | `not` | Boolean NOT |
| 7 | `and` | Boolean AND |
| 8 (lowest) | `or` | Boolean OR |

Use **parentheses** to override precedence and make your intentions clear.

````python
# Precedence examples
print(2 + 3 * 4)       # 14  (not 20)
print((2 + 3) * 4)     # 20
print(2 ** 3 + 1)      # 9   (not 16)
print(10 % 3 + 1)      # 2   (10%3=1, then 1+1=2)
````

<hr>

Methods are functions that belong to a class, functions can be on any other scope of the code so you could state that all methods are functions, but not all functions are methods.

In this workbook, we have covered several functions:
- `print()`
- `type()`
- `len()`
- Variable Casting Functions

To clarify:
- Function: Is independent of any object. Standalone feature or functionality.
- Method: Is a feature within an object (class).

Different types of objects in Python have different **attributes** that can be referred to by name (similar to a variable). To access an attribute of an object, use a dot (`.`) after the object, then specify the attribute (i.e. `obj.attribute`)

When an attribute of an object is a callable, that attribute is called a **method**. It is the same as a function, only this function is bound to a particular object.

When an attribute of an object is not a callable, that attribute is called a **property**. It is just a piece of data about the object, that is itself another object.


**Some methods on string objects**

- **`.capitalize()`** to return a capitalized version of the string (only first char uppercase)
- **`.upper()`** to return an uppercase version of the string (all chars uppercase)
- **`.lower()`** to return an lowercase version of the string (all chars lowercase)
- **`.count(substring)`** to return the number of occurences of the substring in the string
- **`.startswith(substring)`** to determine if the string starts with the substring
- **`.endswith(substring)`** to determine if the string ends with the substring
- **`.replace(old, new)`** to return a copy of the string with occurences of the "old" replaced by "new"

# Section 7: The `print()` Function Revisited

The `print()` function prints the values to a stream, or to sys.stdout by default.

As seen previously, The `print()` function prints all of its arguments as strings, separated by spaces and follows by a linebreak:

    - print("Hello World")
    - print("Hello",'World')
    - print("Hello", <Variable>)

Note that `print` is different in old versions of Python (2.7) where it was a statement and did not need parentheses around its arguments.

````python
print("Hello","World")
````

The print function has some optional arguments to control where and how to print. This includes:
- `file` >> a file-like object (stream); defaults to the current sys.stdout.
- `sep` >> string inserted between values, default a space.
- `end` >> string appended after the last value, default a newline.
- `flush` >> whether to forcibly flush the stream.

When writing to a file, setting the argument `flush=True` may be useful to force the function to write the output immediately. Without this, Python may buffer the output which helps to improve the speed for repeated calls to print(), but isn't helpful if you are, for example, wanting to see the output immediately during debugging.

Let's see some of these optional arguments in an example:

````python
print("Hello","World",sep='...',end='!!',flush=True)
````

### f-Strings (Formatted String Literals)

**f-strings** (introduced in Python 3.6) are the most modern and readable way to embed expressions inside strings. Prefix the string with `f` or `F` and wrap variables/expressions in curly braces `{}`.

```python
name = "Alice"
age = 30
print(f"My name is {name} and I am {age} years old.")
# My name is Alice and I am 30 years old.
```

f-strings support expressions and formatting:
```python
pi = 3.14159
print(f"Pi to 2 decimal places: {pi:.2f}")   # Pi to 2 decimal places: 3.14
print(f"10 squared is {10 ** 2}")             # 10 squared is 100
```

````python
city = "London"
temperature = 18.5

# f-string
print(f"The temperature in {city} is {temperature} degrees.")

# With formatting specifier
price = 1234.5678
print(f"Formatted price: £{price:,.2f}")
````
