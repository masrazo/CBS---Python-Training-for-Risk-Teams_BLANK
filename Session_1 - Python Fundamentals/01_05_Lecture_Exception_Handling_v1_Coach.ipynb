# Section 1: What Are Exceptions?
---

When Python encounters a problem it cannot continue from, it **raises an exception** -- an object that describes the error.

If the exception is not handled, the program stops and prints a **traceback**:

```
ZeroDivisionError: division by zero
```

**Errors vs Exceptions:**

| Type | What it is | Example |
|------|------------|--------|
| **SyntaxError** | Code that Python cannot parse -- caught before running | `if x =` |
| **Exception** | A problem that occurs during execution -- can be caught and handled | `1 / 0` |

Almost everything you will handle in practice is an **Exception**.

# Section 2: Your First `try` / `except`
---

The simplest form of exception handling looks like this:

```python
try:
    # code that might fail
    ...
except:
    # what to do if it does
    ...
```

**Flow:**
1. Python runs the `try` block
2. If anything goes wrong -> jumps immediately to `except`
3. If nothing goes wrong -> `except` is skipped
4. Execution continues after the `try/except`

# Section 3: Common Exception Types
---

Python has many built-in exception types. Here are the ones you will see most often:

| Exception | When it occurs | Example |
|-----------|---------------|--------|
| `TypeError` | Wrong type for an operation | `'a' + 1` |
| `ValueError` | Right type, invalid value | `int('hello')` |
| `ZeroDivisionError` | Dividing by zero | `10 / 0` |
| `IndexError` | Sequence index out of range | `lst[99]` on a 3-element list |
| `KeyError` | Dictionary key not found | `d['missing']` |
| `AttributeError` | Object has no such attribute | `'hello'.nonexistent` |
| `NameError` | Variable not defined | Using `x` before assigning it |
| `FileNotFoundError` | File does not exist | `open('missing.csv')` |
| `RecursionError` | Recursion depth exceeded | Infinite recursive call |

Each of these is a **subclass** of `Exception`. You will see in Section 4 why that matters.

# Section 4: Catching Specific Exceptions
---

Catching specific exception types lets you respond differently to different problems -- and crucially, **avoids hiding bugs you didn't expect**.

```python
try:
    ...
except TypeError:
    # handle TypeError
except ValueError:
    # handle ValueError
except (IndexError, KeyError):
    # handle either, the same way
```

> **Always catch the most specific exception first.** If you catch `Exception` before `ZeroDivisionError`, the `ZeroDivisionError` handler will never run.

````python
##We have a script that divides 1 by the elements of a list:
for i in [2,1.5,0.0,3, "hola"]:
    inverse = 1.0/i
    print(i, inverse)
````

````python
##Catch the exception in general:
for i in [2,1.5,0.0,3, "hola"]:
    try:
        inverse = 1.0/i
        print(i, inverse)
    except:
        print("error with value: ", i)
````

````python
##Adding different behaviour depending on the error
##if there is a zero division error, then divide by 1
##if there is a type error, divide by the string len 
##Catch any other type of error
for i in [2,1.5,0.0,3, "cow"]:
    try:
        inverse = 1.0/i
        print(i, inverse)
    except ZeroDivisionError:
        inverse = 1.0/1
        print("Zero Division Error",1, inverse)
    except TypeError:
        inverse = 1.0 / len(i)
        print("Type Error",len(i), inverse)
````

````python
##The following is a similar funtion just get the index of the list, this to explore a third type of error "IndexError"
def process_input(value, index):
    data = [10, 20, 30]
    try:
        number = int(value)
        return data[index] / number
    except ValueError:
        print(f'ValueError: {value!r} is not an integer')
    except IndexError:
        print(f'IndexError: index {index} out of range')
    except ZeroDivisionError:
        print('ZeroDivisionError: division by zero')

process_input('5',   0)
process_input('abc', 0)
process_input('5',  10)
process_input('0',   1)
````

# Section 5: Raising Custom Exceptions
---

Use `raise` to intentionally trigger an exception -- typically to enforce preconditions or signal that something is wrong with the inputs.

```python
def set_age(age):
    if not isinstance(age, int):
        raise TypeError(f'age must be an int, got {type(age).__name__}')
    if age < 0 or age > 150:
        raise ValueError(f'age must be between 0 and 150, got {age}')
    return age
```

**Guidelines for raising exceptions:**
- Use a **specific** exception type, not a bare `Exception`
- Write a **helpful message** -- include the bad value and what was expected
- Raise **early** -- validate inputs at the start of a function

````python
def set_age(age):
    if not isinstance(age, int):
        raise TypeError(f'age must be an int, got {type(age).__name__}')
    if not 0 <= age <= 150:
        raise ValueError(f'age must be 0-150, got {age}')
    return age

for bad in ['thirty', -5, 200, 30]:
    try:
        set_age(bad)
        print(f'OK: {bad}')
    except (TypeError, ValueError) as e:
        print(f'{type(e).__name__}: {e}')
````

# Section 6: Best Practices
---

| Practice | Why |
|----------|-----|
| **Catch specific exceptions** | Bare `except:` hides real bugs |
| **Keep `try` blocks short** | Only wrap the code that might raise |
| **Write helpful messages** | Include the bad value and what was expected |
| **Don't silence exceptions** | `except: pass` is almost always wrong |
| **Raise early, fail loudly** | Validate inputs at the start |
| **Use `finally` for cleanup** | Or better: use `with` statements |
| **Log exceptions in production** | Use Python's `logging` module rather than `print` |

**Anti-patterns to avoid:**

```python
# BAD -- catches everything, hides bugs
try:
    do_something()
except:
    pass

# BAD -- too broad
try:
    a = load_data()
    b = process(a)
    c = save(b)
except Exception:
    print('Something failed')

# GOOD -- specific, short, informative
try:
    a = load_data()
except FileNotFoundError as e:
    raise DataLoadError(f'Could not load source: {e}') from e
```

def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print(f'Warning: cannot divide {a} by zero')
        return None
    except TypeError as e:
        print(f'Warning: invalid types -- {e}')
        return None
    else:
        print(f'{a} / {b} = {result}')
        return result
    finally:
        print('safe_divide() complete')

import pandas as pd
def safe_read_csv(filepath, required_columns):
    try:
        df = pd.read_csv(filepath)
    except FileNotFoundError:
        raise FileNotFoundError(f'CSV file not found: {filepath!r}')
    missing = [c for c in required_columns if c not in df.columns]
    if missing:
        raise ValueError("Columns missing: ", missing)
