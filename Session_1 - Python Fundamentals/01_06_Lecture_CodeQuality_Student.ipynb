## A motivating example

Here is a function that takes a list of prices, applies a discount, and adds tax. It runs and gives the right answer, but it is hard to read and hard to trust. We will improve it step by step through this notebook.

````python
# BEFORE: it works, but would you want to maintain it?
def f(p, d):
    res = []
    for i in p:
        res.append(i * (1 - d) * 1.2)   # 1.2 = add 20% tax after the discount
    return res

f([100, 200, 300], 0.1)
````

**What makes this hard to maintain?**

- The name `f` tells you nothing about what it does.
- The parameters `p` and `d` give no clue what to pass in.
- `1.2` is a *magic number*: you only learn it means tax by reading the comment.
- There is no docstring, so a user cannot tell how to call it without reading the body.
- There are no type hints, so it is unclear `p` is a list and `d` a fraction.
- Nothing checks the inputs, so a bad value produces a silently wrong result.

Over the next sections we fix each of these. The fully cleaned-up version appears at the end of the maintainability part, so you can compare it with the code above.

    ...

# Section 6: Keep configuration out of your code

Hardcoding settings (file paths, rates, thresholds, credentials) scatters them through your code: they are hard to find, hard to change, and unsafe to share. Keep them in one place. A common pattern is a separate **`config.py`** module that holds the settings, which your code then imports. This also helps reproducibility and safety: you can swap settings for a different environment without touching the logic, and secrets stay out of the shared notebook.

### A `config.py` example

```python
# config.py
DATA_DIR = "../datasets"
DISCOUNT_RATE = 0.1
TAX_RATE = 0.2
```

and your notebook simply imports what it needs:

```python
from config import DATA_DIR, TAX_RATE
```

This project already uses the pattern: the Session 2 notebooks import database settings with `from config import db_params`, and `config.py` is kept out of version control (gitignored) because it holds credentials.

# Section 8: Notebook hygiene

A notebook remembers variables from cells you have since edited or deleted (*hidden state*), so a notebook that works for you can fail for someone running it fresh. The single most important habit:

> **Before sharing, run `Kernel > Restart & Run All`** and confirm it works top to bottom.

Also: keep cells in execution order, do not rely on cells run out of sequence, and clear noisy output before sharing. These points are the core of the *Ten Simple Rules for Reproducible Research in Jupyter Notebooks*.

<a id="takeaways"></a>
## Key Takeaways

Use this as a checklist before you share any notebook.

| Habit | Why it matters |
|---|---|
| Clear `snake_case` names | Code reads like prose; fewer comments needed |
| Docstrings (why, not what) | Others know how to use your function |
| DRY + named constants | Change a value in one place; no magic numbers |
| Type hints | Self-documenting; tools catch mistakes |
| Validate inputs | Fail loudly and early, not silently wrong |
| Configuration in one place (config.py) | Easy to change; secrets stay out of code |
| Relative paths | Runs on any machine |
| Restart & Run All | No hidden-state surprises |

## Conclusion

Maintainable, reproducible code is a habit, not a one-off. Apply a little of this to everything you write and your work becomes easier to trust, review and reuse.
