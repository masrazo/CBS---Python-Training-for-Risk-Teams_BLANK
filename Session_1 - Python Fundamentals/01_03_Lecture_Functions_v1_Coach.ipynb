## Defining a Function  

```python
def funcname(arg1, arg2, ... argN):
    ''' Documentation String '''
    statements
    return value
```

### Understanding the Syntax:

- A function named `"funcname"` is defined.
- It accepts arguments `"arg1, arg2, ..., argN"`.
- The function includes a **docstring** (`'''Documentation String'''`) to describe its purpose.
- It executes a series of **statements**.
- The function **returns a value** (optional). If no `return` statement is used, Python automatically returns `None`, which is equivalent to `False` in a Boolean context.

Using functions properly makes your code **structured, maintainable, and efficient**.

Use **Google-style docstrings** to document Python functions clearly and consistently. This format is widely adopted and supported by documentation tools such as **Sphinx (via the Napoleon extension)**.

### Core principles

- Place the docstring **immediately after the `def` statement**.
- Start with a **short, one-line summary** written in the **imperative mood** (e.g. "Calculate…", "Return…").
- Leave a **blank line** after the summary.
- Document the **behaviour and intent**, not the implementation.
- Use consistent section headers such as `Args`, `Returns`, and `Raises`.
- Keep descriptions concise and precise.

### Standard structure [Google style](https://google.github.io/styleguide/pyguide.html )

# Section 3: Returning Multiple Values
---

**Multiple variables can also be returned as a tuple from a function**. However, returning too many values in this way can reduce readability and may lead to errors when the order of the returned values is misinterpreted.

For example, if a function is called without any variables to assign the returned values to, the result will be returned as a tuple. But, if variables are specified, the values are assigned to those variables in the order they appear in the return statement.

```python
def example_func():
    return 1, 2, 3

# Calling the function without assigning variables
result = example_func()
print(result)  # Output: (1, 2, 3)

# Calling the function with variables to assign the returned values
a, b, c = example_func()
print(a, b, c)  # Output: 1 2 3
```

In this case:

* The function `example_func()` returns a tuple `(1, 2, 3)`.
* When calling the function without variables, the result is stored as a tuple.
* When variables are specified, the values are unpacked and assigned to a, b, and c in the specified order.

Lets test it out!

````python
eglist = [10, 50, 30, 12, 6, 8, 100]
````

## 4.1 Positional Arguments -- The Basics

Arguments/parameters are values passed to a function to customise its behaviour and enable it to perform operations on different inputs. The simplest functions take one or more **positional arguments**. Position matters -- the first value passed becomes the first parameter.

````python
# Simplest: one positional argument
def greet(name):
    print(f'Hi {name}!')

greet('Alice')
greet('Bob')
````

````python
# Two positional arguments -- order matters
def divide(numerator, denominator):
    return numerator / denominator

print(divide(10, 2))   # 5.0
print(divide(2, 10))   # 0.2  -- different order, different result
````

## 4.2 More Complex Examples

### **Implicit/Default Arguments**

When an argument of a function is common in the majority of cases, this can be specified with a default value. This is also called an implicit argument.

For example:

```python
def implicitadd(a, b=3, c=0):
    return a + b + c
```

In this function, `implicitadd()` accepts up to three arguments, but most of the time, the second argument is added by 3, and the third argument is 0. Therefore, the last two arguments are default arguments.

Now, if the second argument is not defined when calling the `implicitadd()` function, it is considered to be 3.

However, we can call the same function with two or three arguments. A useful feature is to explicitly name the argument values being passed into the function. This gives great flexibility in how to call a function with optional arguments. All of the following are valid:

```python
print(implicitadd(5))           # Uses default b=3, c=0. Output: 8
print(implicitadd(5, 4))        # Uses default c=0. Output: 9
print(implicitadd(5, 4, 2))     # No defaults, all values specified. Output: 11
print(implicitadd(a=5, c=2))    # Explicitly naming arguments. Output: 10
```

A **default argument** lets you make a parameter **optional**. If the caller doesn't supply a value, the default is used.

````python
# Simple default argument
def greet(name='World'):
    print(f'Hello, {name}!')

greet()           # Hello, World!
greet('Alice')    # Hello, Alice!
````

## 4.4 Variable Arguments -- `*args` and `**kwargs`

What if you don't know in advance **how many arguments** the caller will pass? Python gives you two special syntaxes:

- **`*args`** -- collects extra **positional** arguments into a **tuple**
- **`**kwargs`** -- collects extra **keyword** arguments into a **dictionary**

The names `args` and `kwargs` are conventions -- the magic is in the `*` and `**` prefixes.

### 4.4.1 `*args` -- Variable Number of Positional Arguments

Use `*args` when the function should accept **any number** of positional values.

Inside the function, `args` behaves like a regular **tuple**.

````python
def sum_all(*args):
    print(f'args is a {type(args).__name__}: {args}')
    return sum(args)

print(sum_all(1, 2, 3))            # 6
print(sum_all(10, 20))             # 30
print(sum_all(5))                  # 5
print(sum_all())                   # 0  -- works with zero arguments
````

````python
# Combine fixed and variable positional arguments
def log_event(level, *messages):
    """level is required; any number of messages can follow."""
    for msg in messages:
        print(f'[{level.upper()}] {msg}')

log_event('info',  'Server started')
log_event('error', 'Disk full', 'Cannot write log', 'Retrying...')
````

### 4.4.2 `**kwargs` -- Variable Number of Keyword Arguments

Use `**kwargs` when the function should accept **any number of named arguments**.

Inside the function, `kwargs` behaves like a regular **dictionary**.

````python
def build_profile(**kwargs):
    print(f'kwargs is a {type(kwargs).__name__}: {kwargs}')
    for key, value in kwargs.items():
        print(f'  {key}: {value}')

build_profile(name='Alice', age=30, city='London')
````

````python
# Common pattern: passing config-like options
def make_chart(data, **options):
    title  = options.get('title',  'Untitled')
    colour = options.get('colour', 'blue')
    grid   = options.get('grid',   True)
    print(f'Chart "{title}" with {len(data)} points, colour={colour}, grid={grid}')

make_chart([1,2,3])
make_chart([1,2,3], title='Sales', colour='red')
make_chart([1,2,3], title='Q1', grid=False, colour='green')
````

### 4.4.3 Argument Order Rules

When mixing argument types, the **order matters**. The valid order is:

1. positional arguments
2. default arguments
3. `*args`
4. keyword-only arguments
5. `**kwargs`

```python
def f(positional, default=0, *args, keyword_only, **kwargs):
    ...
```

````python
def describe_user(name, age=18, *roles, country, **extra):
    """Show off every type of argument in one function."""
    print(f'{name}, age {age}, from {country}')
    if roles:
        print(f'  Roles: {", ".join(roles)}')
    for k, v in extra.items():
        print(f'  {k}: {v}')

describe_user('Alice', 30, 'admin', 'developer', country='UK', team='Data')
describe_user('Bob', country='IE')   # minimal call
````

### 4.4.4 Unpacking at the Call Site

The same `*` and `**` operators work in **reverse** when calling a function -- they **unpack** a list/tuple or dict into individual arguments.

This is incredibly useful for forwarding arguments, building flexible wrappers, or when you already have your data in a collection.

````python
def make_point(x, y, z):
    return f'({x}, {y}, {z})'

# Without unpacking
print(make_point(10, 20, 5))

# With *list -- unpacks each element as a positional argument
coords = [10, 20, 5]
print(make_point(*coords))   # same as make_point(10, 20, 5)

# With **dict -- unpacks key/value as keyword arguments
kw = {'x': 10, 'y': 20, 'z': 5}
print(make_point(**kw))      # same as make_point(x=10, y=20, z=5)
````

````python
# Real-world pattern: forwarding kwargs to another function
def log_and_run(func, *args, **kwargs):
    print(f'Calling {func.__name__} with args={args}, kwargs={kwargs}')
    return func(*args, **kwargs)

def add(a, b): return a + b
def greet(name, polite=True):
    return f'Hi {name}.' if polite else f'Hey {name}!'

print(log_and_run(add, 5, 7))
print(log_and_run(greet, 'Alice', polite=False))
````

````python
def askName():
    attempts = 3
    while attempts > 0:
        nInput = input("Please Enter Your Name: ")
        if nInput.isalpha():
            print("Input Accepted\n")
            return nInput
        attempts-=1
        print(">> Please Enter Alphabetical Characters Only\n>> %i attemps remaining\n" % attempts)
    else:
        print("Input Rejected")
        return ""
````

## Global & Local Variables

Whatever variable is declared inside a function is a **local** variable and outside the function is a **global** variable.

There is an important distinction between the two.

Global variables can be accesed from anywhere in the program whilst local variables are only accessible within the function or block where they are defined.

Take the example below. The variable `x` is assigned different values throughout the code, notice which are global and local:

````python
def func1():
    x = 1
    def func2():
        x = 2
        print("Local func2 x =", x)
    func2()
    print("Local func1 x =", x)
````

### __Example: Currency Conversion__

For this example, we want to convert a list of prices in British pounds (GBP) to US dollars (USD) and round them. Without using modular code, it would look like this:

````python
# Example: Without modular code
prices_gbp = [10, 20, 30, 50]
prices_usd = []
exchange_rate = 1.24
for price in prices_gbp:
    converted = price * exchange_rate
    rounded = round(converted, 2)
    prices_usd.append(rounded)
print(prices_usd)
````

Here we have another list we need to apply the conversion.

````python
# Another list to convert from GBP to USD: 
services_gbp = [15, 25, 60]
services_usd = []
exchange_rate = 1.24
for price in services_gbp:
    converted = price * exchange_rate
    rounded = round(converted, 2)
    services_usd.append(rounded)
print(services_usd)
````

Using modular code modular code to create a currency converter, we will define two functions `convert_gbp_to_usd` and `batch_convert`. The first function converts a price in GBP to USD by an exchange rate. The second function bulk processes the conversion of a price list in GBP to USD. This design promotes code reusability and provide flexibility through configurable exchange rates.

````python
# Funtion 1: GBP to USD
def convert_gbp_to_usd(gbp_amount, exchange_rate = 1.24):
    """Convert an amount in British Pounds (GBP) to US Dollars (USD).

    This function multiplies the given GBP amount by a exchange rate
    and returns the result rounded to two decimal places.

    Args:
        gbp_amount: A float representing an amount in GBP.
        exchange_rate: A float representing the exchange rate from GBP Pound to USD dollar

    Returns:
        A float representing the equivalent amount in USD.
    """
    return round(gbp_amount * exchange_rate, 2)
````

Let's test our first function to convert a single element of our prices list into USD.

````python
# Testing conversion function on a single element on our list: 
prices_gbp = [10, 20, 30, 50]
print(convert_gbp_to_usd(prices_gbp[0]))
````

Now, onto our second function which will apply the conversion to the entire list and return a list of prices in USD.

````python
# Function 2: Converting list of prices
def batch_convert(prices):
    """Convert a list of prices from GBP to USD using a helper function.

    Applies `convert_gbp_to_usd` to each element in the list.

    Args:
        prices: A list of float values in GBP.

    Returns:
        A list of float values converted to USD.
    """
    #return values in USD.
    return [convert_gbp_to_usd(p) for p in prices]
````

Let's call the `batch_convert` function to convert out GBP prices in the lists `prices_gbp` and `services_gbp` to USD.

````python
# Converting lists to USD
print(batch_convert(prices_gbp))
print(batch_convert(services_gbp))
````

### __Example: Sentiment Tagging on Tweets__

In this section, we'll explore a simplified form of sentiment analysis, a common NLP task in data science. The goal is to classify tweets into 'positive', 'negative', or 'neutral' based on a simple keyword detection. This is a very limited and rule-based approach. In practice, more advanced methods like logistic regression, transformers, or pretrained language models (e.g., BERT, RoBERTa) are used to capture context and semantics.

Step-by-step logic for this task:
1. Clean the tweets by removing symbols like hashtags and mentions, and convert to lowercase.
2. Define a set of keywords that reflect positive, neutral, and negative sentiments.
3. For each tweet:
    - Count how many positive, neutral, and negative words it contains.
    - If the number of positive words > neutral, classify as 'positive'.
    - If the number of negative words > both neutral and positive, classify as 'negative'.
    - Otherwise, classify as 'neutral'.
4. Tally the counts for each sentiment class.
5. Use modular functions to make the logic reusable and clean.

__We are going to follow a progressive approach, focusing on 'positive' and 'neutral' classes for now.__

Below, we are labelling the sentiment of the tweets without using modular code.

````python
# Example: Non modular 
# Tweets:
raw_tweets = [
    "@user I LOVE this! #awesome",
    "Not bad, but not great either.",
    "@brand Happy to see this update!",
]

# Dictionary of positive and negative classes:
positive_words = ["great", "love", "awesome", "nice", "happy"]
neutral_words = ["ok", "fine", "meh", "average", "not bad", "decent"]

# List that will store the classes of each tweet. 
labels = []
for tweet in raw_tweets:
    # Cleaning the tweet: 
    text = tweet.lower().replace("#", "").replace("@", "").strip()
    # Counting posstive and neutral word on the tweet:
    pos_count = sum(1 for word in positive_words if word in text)
    neut_count = sum(1 for word in neutral_words if word in text)
    # Labeling the string depending on the possitive and neutral instances:
    if pos_count > neut_count:
        labels.append("positive")
    else:
        labels.append("neutral")

print(labels)
````

Using modular code, we can define three complimentary functions to follow our logic. The first function, `clean_text`, will clean and standardise the tweet text by removing symbols and converting all characters to lower case. The second function, `count_matching_words`, will match words to a list of sentiment keywords and count the number of occurences. The third and main function, `tag_sentiment`, coordinates preprocessing and analysis to deliver sentiment classiciation counts.

````python
# Modular form:

def clean_text(text):
    """Clean tweet text by removing symbols and converting to lowercase.

    This function removes hashtags and mentions, trims whitespace,
    and converts all characters to lowercase.

    Args:
        text: A string containing a tweet.

    Returns:
        A cleaned string suitable for text analysis.
    """
    return text.lower().replace("#", "").replace("@", "").strip()

def count_matching_words(text, word_list):
    """Count how many words from the word_list appear in the given text.

    Args:
        text: A cleaned string.
        word_list: A list of sentiment keywords to match.

    Returns:
        An integer count of how many words from word_list are found in text.
    """
    return sum(1 for word in word_list if word in text)

def tag_sentiment(text):
    """Tag sentiment of a tweet as positive or neutral.

    Compares number of positive and neutral words to assign a sentiment.

    Args:
        text: A raw tweet string.

    Returns:
        A string label: 'positive' or 'neutral'.
    """
    cleaned = clean_text(text)
    pos_count = count_matching_words(cleaned, positive_words)
    neut_count = count_matching_words(cleaned, neutral_words)
    return "positive" if pos_count > neut_count else "neutral"
````

Now, let's find the count of positive and neutral tweets in the list.

````python
# Count how many tweets are positive vs neutral
sentiment_counts = {"positive": 0, "neutral": 0}
for tweet in raw_tweets:
    # Tag the sentiment of each tweet
    sentiment = tag_sentiment(tweet)
    # Update the corresponding sentiment counter
    sentiment_counts[sentiment] += 1

print(sentiment_counts)
````

# Section 9: Recursion
---

A **recursive** function calls itself. Every recursive function needs:

1. **Base case** -- the condition that stops the recursion
2. **Recursive case** -- a call to itself, moving closer to the base case

> Without a base case, a recursive function will call itself until Python raises a `RecursionError`.

***

## Recursion

Recursion, in the context of computer science and by extension data science, refers to a method where the solution to a problem depends on solutions to smaller instances of the same problem. This technique is characterised by a function calling itself, either directly or indirectly, until a specific base condition is met. Grasping recursion is fundamental for understanding the architecture of many advanced algorithms prevalent in machine learning and data processing.

__Core Principles of Recursion:__

Every well-formed recursive function adheres to two critical components:

- __Base Case:__ This is the terminal condition that halts the recursion. Without a clearly defined base case, the function would call itself indefinitely, leading to a RecursionError (or "stack overflow") due to exhaustion of the call stack. The base case represents the simplest instance of the problem that can be solved directly without further recursive calls.

- __Recursive Step:__ This is the part of the function where it invokes itself with a modified input. The modification must ensure that each successive call moves closer to the base case, thereby progressively simplifying the problem until it can be resolved by the base case.

### __Example - Russian Nesting Dolls__

Imagine a set of Russian nesting dolls, each containing a smaller version of itself, until you reach the smallest doll which contains nothing. To open all the dolls, you'd pick one up, open it, and then repeat the process for the doll inside. We are going to simulate opening all the dolls using a function.

````python
# Example: Russian dolls
def open_russian_dolls(doll_size):
    """
    Simulates the process of opening Russian nesting dolls recursively.

    Args:
        doll_size: The current size of the doll to be opened.
                   Assumes doll_size is a positive integer.
    """
    print(f"Opening a doll of size: {doll_size}")

    # Base Case: The smallest doll (size 1) contains nothing further.
    if doll_size == 1:
        print("Found the smallest doll! No more dolls inside.")
        return # Terminates the recursion for this branch

    # Recursive Step: Open the current doll, then proceed to open the smaller doll inside.
    else:
        open_russian_dolls(doll_size - 1) # Recursive call with a decremented doll size

# Let's begin with a larger doll!
open_russian_dolls(11)
````

__Analysis:__

- **Base Case**: When `doll_size` is 1, the function prints a message and returns, stopping the recursive calls for that particular line of execution.

- **Recursive Step**: For any `doll_size` greater than 1, the function calls `open_russian_dolls` with `doll_size - 1`, ensuring that each call moves closer to the base case of `doll_size == 1`.

### __Example - The Gossip Chain:__

Imagine you need to spread gossip through a social network, but each person can only tell one other person. Using the function below, we will be spreading gossip to a group of people.

````python
# Example: The gossip chain
def spread_gossip(people_list, gossip, source="original source"):
    """
    Recursively spreads gossip through a list of people, tracking the source.
    
    Args:
        people_list: A list of strings representing people's names.
        gossip: A string containing the gossip message to spread.
        source: A string representing who told the gossip to the current person.
        
    Returns:
        None. Prints the gossip spreading process with source tracking.
    """
    # Base case: no more people to tell
    # When the list is empty, we've reached everyone
    if not people_list:
        print("Gossip has reached everyone!")
        return
    
    # Recursive case: tell the first person, then continue with the rest
    # Extract the current person (first in list)
    current_person = people_list[0]
    # Create a new list without the current person for the next recursive call
    remaining_people = people_list[1:]
    
    # Show who heard the gossip and from whom
    print(f"{current_person} heard from {source}: '{gossip}'")
    
    # Recursively call the function with:
    # - remaining people (smaller problem)
    # - same gossip message
    # - current person becomes the source for the next person
    spread_gossip(remaining_people, gossip, current_person)
````

Let's use the function to spread gossip to a list of friends.

````python
# Spreading gossip
friends = ["Alice", "Bob", "Charlie", "Diana"]
spread_gossip(friends, "There are free biscuits in the kitchen!")
````

### __Key Applications of Recursion in Data Science__

Recursion is a powerful problem-solving technique where a function solves a problem by calling itself, typically breaking it down into smaller, self-similar sub-problems until a simple **base case** is reached. This fundamental concept is essential for understanding several areas in data science:

1. Tree-Based Machine Learning Models: Algorithms like **Decision Trees**, and by extension **Random Forests** and **Gradient Boosting Machines**, are inherently recursive. They build a model by repeatedly splitting data at each node until a stopping condition is met (e.g., maximum depth, minimum samples). Understanding this recursive nature is crucial for comprehending how these widely used predictive models are constructed and how hyper-parameters like `max_depth` influence their behavior.

2. Graph and Network Analysis: Many algorithms for exploring connections within networks, such as **Depth-First Search (DFS)**, leverage recursion. They navigate through a graph by diving deep into each branch before backtracking. Useful for tasks like identifying connected components, finding paths, or analyzing relationships in social networks or recommendation systems.

3. Processing Nested Data Structures: Recursion provides an elegant and efficient way to handle data in hierarchical formats like **JSON** or **XML**. It allows you to parse, navigate, and extract information from objects that can be nested at arbitrary depths. Simplifies working with complex, non-tabular data formats often encountered in web scraping, API responses, or NoSQL databases.


In essence, recursion helps you grasp how complex data structures are processed and how many sophisticated machine learning algorithms function at their core.

players = ["Mario", "Luigi", "Peach", "Yoshi", "Bowser", "Toad"]

def display_player_positions():
    """Display player rankings in a readable format."""
    print("\n🏁 Current Player Positions 🏁")
    for i, player in enumerate(players, start=1):
        position = f"{i}{'st' if i==1 else 'nd' if i==2 else 'rd' if i==3 else 'th'}"
        print(f"{position}: {player}")
    print("-" * 30)  # Separator for readability

def get_shell_choice():
    """Prompt the user to choose a shell and validate input."""
    while True:
        shell = input("Choose a shell to launch (red/blue) or type 'quit' to exit: ").lower()
        if shell in ["red", "blue", "quit"]:
            return shell
        else:
            print("Invalid choice. Please enter 'red', 'blue', or 'quit'.")

def get_player_position():
    """Prompt the user to choose a player position and validate input."""
    while True:
        try:
            player_input = int(input(
                f"Enter the position of the player launching the shell (1-{len(players)}): ")) - 1
            
            if 0 <= player_input < len(players):
                return player_input
            else:
                print(f"Invalid position. Choose a number between 1 and {len(players)}.")
        except ValueError:
            print("Invalid input. Please enter a number.")

        target_red_shell = player_position - 1
        print(f"\n🎯 The Red Shell hits {players[target_red_shell]}!")
        # Move the target player back one position
        players[target_red_shell], players[player_position] = players[player_position], players[target_red_shell]

        print("\n❌ No player ahead to target!")

    """Handle the logic for launching a blue shell."""
