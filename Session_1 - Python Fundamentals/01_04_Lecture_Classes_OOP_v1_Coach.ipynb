OOP models the real world as a collection of **objects**, each combining:

- **Attributes** -- data the object holds (e.g. a dog's name, breed, age)
- **Methods** -- actions the object can perform (e.g. bark, sit, fetch)

A **class** is the blueprint. An **instance** (object) is a specific realisation of that blueprint.

```
Class: Dog          <-  blueprint (defined once)
  attributes: name, breed, age
  methods:    bark(), fetch(), sit()

Instance: my_dog = Dog('Rex', 'Labrador', 3)   <-  specific object
```

| Term | Meaning |
|------|---------|
| **Class** | Blueprint / template |
| **Instance / Object** | A specific realisation of a class |
| **Attribute** | Data stored on an object (`self.name`) |
| **Method** | A function defined inside a class |
| **`__init__`** | Called automatically when an object is created |
| **`self`** | Reference to the current instance |

## Basic Syntax

```python
class ClassName:
    def __init__(self, param1, param2):
        self.param1 = param1
        self.param2 = param2

    def method_name(self):
        return self.param1
```

- `class` keyword starts the definition
- `__init__` is the **constructor** -- runs every time you create a new instance
- `self` is always the first parameter; it refers to the instance itself
- Attributes prefixed with `self.` are accessible throughout the class

Python is an "object-oriented programming language." This means that almost all the code is implemented using a special construct called classes. Programmers use classes to keep related things together. This is done using the keyword "class" which is a grouping of object-oriented constructs.

A **class** is a code template for creating objects. Objects have member variables and have behavior associated with them. In Python, a class is created by the keyword class.


A class is declared as follows:

```python
class class_name:
    methods (functions)```

Lets create a class called `Students` below:

````python
class Students:
    school = "python_school"
````

`school` is a class attribute, meaning that every instance and object in this class share this attribute. In the case of `Students`, you can think of all the students being from `"python_school"`.

An **Object** is an instance of a Class. It represents a specific implementation of the class and holds its own data.

Now lets create an object from `Students` called `student1`:



````python
class Students:
    school = "python_school"

# Create an object from the class
student1 = Students()

# Access the class attribute
print(student1.school)
````

Since school is a class attribute, it can be directly accessed through student1 as it is an attricute shared by all objects in the class.

The `__init__` function is used to automatically initialise object attributes in the class when they are created.

Take the example below where `name` and `age` will by included in the object when created:

````python
class Students:
    school = "python_school"

    def __init__(self, name, age):
      self.name = name
      self.age = age

student1 = Students("Matt", 24)

print(student1.name)
print(student1.age)
print(student1.school)
````

The `student1 = Students("Matt", 24)` expression creates an object in the Student class with name="Matt" and age=24 (note that the order of arguments is important).

The `self` parameter is used to reference the current instance of the class, enabling access to the attributes and mothods of the object.

Take the below example:

````python
class Students:
    school = "python_school"

    def __init__(self, name, age):
      self.name = name
      self.age = age

    def learn(self):
      print(f"{self.name} who is " + f"{self.age} is learning about classes!")

student1 = Students("Matt", 24)
student1.learn()
````

The `__str__` method in Python allows you to define a custom string representation of an object.

### Class & Instance Variables

In Python, variables defined in a class can be either class variables or instance variables, and understanding the distinction between them is crucial for object-oriented programming.

**Class Variables** are shared across all instances of a class. It is defined at the class level, outside any methods. All objects of the class share the same value for a class variable unless explicitly overridden in an object.

**Instance Variables** are  unique to each instance (object) of a class. These are defined within `__init__` method or other instance methods. Each object maintains its own copy of instance variables, independent of other objects.

Lets look at the difference between them in an example:

````python
class Students:
    # Class Variable
    school = "python_school"

    def __init__(self, name, age):
      # Instance Variables
      self.name = name
      self.age = age

# Create Objects
student1 = Students("Matt", 24)
student2 = Students("Maria", 30)

# Access class and instance variables
print(student1.school)    # Class variable
print(student1.name)      # Instance variable

# Modify instance variables
student1.name = "Sinead"
print(student1.name)      # Updated instance variable

# Modify class variable
Students.school = "SQL_academy"
print(student1.school)    # Updated class variable
print(student2.school)
````

## Class Attributes vs Instance Attributes

- **Instance attributes** (in `__init__`) belong to each individual object
- **Class attributes** (defined at class level) are shared by all instances

````python
class Dog:
    species = 'Canis familiaris'   # class attribute -- shared

    def __init__(self, name, breed):
        self.name  = name          # instance attribute -- unique
        self.breed = breed

    def bark(self):
        return f'{self.name} says Woof!'

rex   = Dog('Rex',   'Labrador')
bella = Dog('Bella', 'Poodle')

print(rex.species, '|', rex.name,   rex.bark())
print(bella.species, '|', bella.name, bella.bark())
````

# Section 3: Dunder (Magic) Methods
---

Dunder methods (double underscore on each side) define how Python's built-in operations work on your objects.

| Method | Triggered by | Purpose |
|--------|-------------|--------|
| `__init__` | `ClassName()` | Constructor -- initialise the object |
| `__str__` | `print(obj)`, `str(obj)` | Human-readable string |
| `__repr__` | `repr(obj)`, REPL display | Unambiguous developer-friendly string |
| `__len__` | `len(obj)` | Return the length |
| `__eq__` | `obj1 == obj2` | Equality comparison |
| `__lt__` | `obj1 < obj2` | Less-than comparison |
| `__add__` | `obj1 + obj2` | Addition operator |

````python
class Book:
    def __init__(self, title, author, pages):
        self.title  = title
        self.author = author
        self.pages  = pages

    def __str__(self):
        return f'"{self.title}" by {self.author}'

    def __repr__(self):
        return f'Book({self.title!r}, {self.author!r}, {self.pages})'

    def __len__(self):
        return self.pages

    def __eq__(self, other):
        return self.title == other.title and self.author == other.author

    def __lt__(self, other):
        return self.pages < other.pages

b1 = Book('Python Crash Course', 'Eric Matthes', 544)
b2 = Book('Fluent Python', 'Luciano Ramalho', 792)

print(b1)
print(repr(b1))
print(len(b1))
print(b1 == b2)
print(b1 < b2)
print(sorted([b2, b1], key=len))
````

# Section 4: Inheritance
---

**Inheritance** lets a class (child) reuse and extend another class (parent).

```python
class Animal:           # parent class
    ...
class Dog(Animal):      # child class -- inherits from Animal
    ...
```

- `super()` calls the parent class's method
- The child can **override** any parent method
- `isinstance(obj, ClassName)` checks if an object is an instance (including inherited)

**Why use it?**
- Avoids code duplication
- Enforces a consistent interface across related classes
- Models real-world hierarchies naturally

````python
class Animal:
    def __init__(self, name, species):
        self.name    = name
        self.species = species

    def speak(self):
        return f'{self.name} makes a sound.'

    def __str__(self):
        return f'{self.name} ({self.species})'


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, 'Canis familiaris')
        self.breed = breed

    def speak(self):
        return f'{self.name} says Woof!'


class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, 'Felis catus')

    def speak(self):
        return f'{self.name} says Meow!'


animals = [Dog('Rex', 'Labrador'), Cat('Whiskers'), Dog('Buddy', 'Beagle')]
for a in animals:
    print(a, '->', a.speak())

print(isinstance(animals[0], Dog))
print(isinstance(animals[0], Animal))
print(isinstance(animals[0], Cat))
````

Let's build a `PublicBus` class step by step. This example shows how OOP captures both **state** (current passengers, route) and **behaviour** (boarding, alighting).

## __Classes__

While functions are for specific actions, classes are for creating objects that bundle together data and the actions that operate on that data.

From a software development perspective, a class is a blueprint for creating objects (instances).  It defines:

- __Attributes:__ Data associated with the object (variables).

- __Methods:__ Functions that the object can perform.

Classes help us __model real-world entities and their relationships__, leading to more organized and maintainable code, especially for complex systems.

### __Example:  Public Transport Bus__ 

In this example, we aim to model a public transport bus using a Python class. Instead of just storing raw data in variables or lists, we use a class to encapsulate both the state and behavior of a bus.

The goal is to simulate basic bus operations such as:

- Boarding and disembarking passengers

- Tracking location

- Monitoring operational status

This example mirrors real-world objects in code, making it a perfect introduction to __object-oriented programming (OOP)__.

````python
# Example: Public Transport Bus
class PublicBus:
    """
    Represents a single public transport bus in a city's fleet.

    Attributes:
        vehicle_id (str): Unique identifier for the bus (e.g., "BUS-A01").
        route_number (str): The current route the bus is assigned to.
        capacity (int): The maximum passenger capacity of the bus.
        current_passengers (int): The number of passengers currently on board.
        current_location (tuple): A tuple (latitude, longitude) representing the bus's current position.
        is_operational (bool): True if the bus is currently in service, False otherwise.
    """
    def __init__(self, vehicle_id, route_number, capacity, initial_location=(0.0, 0.0)):
        """
        Initializes a new PublicBus instance.

        Args:
            vehicle_id (str): Unique identifier for the bus.
            route_number (str): The initial route number for the bus.
            capacity (int): The maximum passenger capacity.
            initial_location (tuple): Starting latitude and longitude. Defaults to (0.0, 0.0).
        """
        self.vehicle_id = vehicle_id
        self.route_number = route_number
        self.capacity = capacity
        self.current_passengers = 0
        self.current_location = initial_location
        self.is_operational = True

    def board_passengers(self, num_passengers):
        """
        Adds passengers to the bus, up to its capacity.

        Args:
            num_passengers (int): The number of passengers attempting to board.
        """
        if not self.is_operational:
            print(f"Bus {self.vehicle_id} is out of service and cannot board passengers.")
            return

        space_available = self.capacity - self.current_passengers
        if num_passengers <= space_available:
            self.current_passengers += num_passengers
            print(f"{num_passengers} passengers boarded Bus {self.vehicle_id}. Current passengers: {self.current_passengers}")
        else:
            boarded = space_available
            self.current_passengers += boarded
            print(f"Bus {self.vehicle_id} is full. {boarded} passengers boarded. {num_passengers - boarded} could not board.")

    def disembark_passengers(self, num_passengers):
        """
        Removes passengers from the bus.

        Args:
            num_passengers (int): The number of passengers disembarking.
        """
        self.current_passengers = max(0, self.current_passengers - num_passengers)
        print(f"{num_passengers} passengers disembarked Bus {self.vehicle_id}. Current passengers: {self.current_passengers}")

    def update_location(self, new_latitude, new_longitude):
        """
        Updates the bus's current geographical location.

        Args:
            new_latitude (float): The new latitude coordinate.
            new_longitude (float): The new longitude coordinate.
        """
        self.current_location = (new_latitude, new_longitude)
        print(f"Bus {self.vehicle_id} moved to ({new_latitude}, {new_longitude}).")
````

#### Understanding Python Class Example: Public Transport Bus

In this example:

- `PublicBus` is our **class**.
- `__init__` is the **constructor.** It sets up a new `PublicBus` object when it's created.
- `vehicle_id`, `route_number`, `capacity`, `current_passengers`, `current_location`, and `is_operational` are **attributes** (the data associated with each bus).
- `board_passengers`, `disembark_passengers`, and `update_location` are **methods** (functions that perform actions on the bus's data).
- `bus_alpha` and `bus_beta` are **instances** (objects) of the `PublicBus` class. Each is a unique bus with its own specific details and state.

__Creating instances (objects) of the class:__

````python
# Creating instances (objects) of our PublicBus class
bus_alpha = PublicBus("BUS-ALPHA", "Route 101", 60, initial_location=(51.458, -0.970)) 
bus_beta = PublicBus("BUS-BETA", "Route 202", 50, initial_location=(51.450, -0.960))  
````

````python
# Accessing attributes:
print(f"Bus Alpha ID: {bus_alpha.vehicle_id}, Route: {bus_alpha.route_number}, Capacity: {bus_alpha.capacity}, Current Passengers: {bus_beta.current_passengers}")
print(f"Bus Beta ID: {bus_beta.vehicle_id}, Route: {bus_beta.route_number}, Capacity: {bus_beta.capacity}, Current Passengers: {bus_beta.current_passengers}")
````

__Simulating operations:__

````python
# Simulating bus operations:
# Bus Alpha boards passengers
# Previous current_passengers value:
print(f"Bus Alpha Current Passengers: {bus_alpha.current_passengers}")

# Updating the passengers value:
bus_alpha.board_passengers(30)
# This changes the `current_passengers` attribute of `bus_alpha`.

# New current passengers value: 
print(f"Bus Alpha Current Passengers: {bus_alpha.current_passengers}")
````

__Demonstrating capacity limits:__

````python
# Demonstrating capacity limits:
# Bus Beta boards passengers
bus_beta.board_passengers(45)

# `current_passengers` for `bus_beta` increases.
bus_beta.board_passengers(10) # Try to board more than capacity allows
# The `board_passengers` method handles the logic for full capacity.

# Bus Alpha disembarks some passengers
bus_beta.disembark_passengers(5)
# `current_passengers` for `bus_alpha` decreases.
````

class Sensor:

        self.location  = location
        self.readings  = []

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner        = owner
        self.balance      = balance
        self.transactions = []

    def deposit(self, amount):
        if amount <= 0: raise ValueError('Deposit must be positive')
        self.balance += amount
        self.transactions.append(('+', amount, self.balance))

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError(f'Insufficient funds ({self.get_balance()})')
        self.balance -= amount
        self.transactions.append(('-', amount, self.balance))

    def get_balance(self):
        return f'£{self.balance:,.2f}'

    def statement(self):
        print(f'Statement for {self.owner}')
        for t, amt, bal in self.transactions:
            print(f'  {t} £{amt:,.2f}   balance: £{bal:,.2f}')
        print(f'Current balance: {self.get_balance()}')

    def __str__(self):
        return f'BankAccount(owner={self.owner}, balance={self.get_balance()})'

acc = BankAccount('Alice', 1000)
acc.deposit(500); acc.withdraw(200); acc.deposit(75)
acc.statement()
print(acc)

import random

class Player:
    TRACK_LENGTH = 100

    def __init__(self, name, speed=5):
        self.name     = name
        self.lap      = 1
        self.position = 0
        self.items    = []
        self.speed    = speed
        self._speed_boost = 0

    def move(self):
        step = self.speed + self._speed_boost
        self._speed_boost = 0   # one-shot
        self.position += step
        if self.position >= self.TRACK_LENGTH:
            self.position -= self.TRACK_LENGTH
            self.lap += 1

    def pick_item(self, item):
        self.items.append(item)

    def use_item(self):
        if not self.items: return None
        item = self.items.pop()
        if item == 'mushroom':
            self._speed_boost = 3
        return item

    def __str__(self):
        return f'{self.name:<10} (lap {self.lap}, pos {self.position:3d}, items: {self.items})'


class HumanPlayer(Player):
    pass


class ComputerPlayer(Player):
    def move(self):
        if random.random() < 0.2:
            self.pick_item('mushroom')
            self.use_item()
        super().move()


class Race:
    def __init__(self, players, total_laps=3):
        self.players    = players
        self.total_laps = total_laps
        self.round      = 0

    def play_round(self):
        self.round += 1
        for p in self.players:
            p.move()

    def standings(self):
        ranked = sorted(self.players,
                        key=lambda p: (p.lap, p.position),
                        reverse=True)
        print(f'-- Round {self.round} --')
        for i, p in enumerate(ranked, 1):
            print(f'  {i}. {p}')

    def winner(self):
        for p in self.players:
            if p.lap > self.total_laps:
                return p
        return None

    def start(self):
        while self.winner() is None:
            self.play_round()
            self.standings()
        w = self.winner()
        print(f'\nWinner: {w.name}!')


random.seed(42)
race = Race([
    HumanPlayer('Mario'),
    ComputerPlayer('Bowser'),
    ComputerPlayer('Peach'),
], total_laps=3)
race.start()
